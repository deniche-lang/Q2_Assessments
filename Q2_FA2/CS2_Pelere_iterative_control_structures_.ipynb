## Part 1: Why Do We Need Loops?

Imagine taking attendance for a class of **30 students**. Without a loop, you might write the same instruction many times:

```python
print("Check student 1")
print("Check student 2")
print("Check student 3")
# ... repeated until student 30
```

This is **code duplication**: the same kind of instruction is copied repeatedly. A loop lets Python repeat instructions for us, making code shorter and easier to update.

````python
# Manual repetition: a few examples of duplicated instructions
print("Check student 1")
print("Check student 2")
print("Check student 3")
````

````python
# Automated repetition: one instruction repeated 30 times
for student_number in range(1, 31):
    print("Check student", student_number)
````

**Think about it:** If the class has 35 students tomorrow, which version is easier to update? With the loop, you only change `31` to `36` in `range(1, 31)`.

A **loop** repeats a block of code. The two common loop structures in Python are `while` and `for`.

## Part 2: `while` Loops and Python's `while-else`

A `while` loop repeats **while its condition is `True`**. It is useful when we do not know in advance how many attempts or repetitions will be needed.

### Example A: Password verification

This classroom example allows repeated attempts until the password is correct. The demo password is `python8`.

````python
correct_password = "python8"
attempt = ""

while attempt != correct_password:
    attempt = input("Enter the demo password (hint: python8): ")
    if attempt != correct_password:
        print("Not quite. Try again.")

print("Access granted!")
````

### Example B: `while-else`

Python has a special `while-else` clause. The `else` block runs when the loop ends **normally because its condition becomes false**. It does **not** run if the loop is stopped by `break`.

Run this example and observe the message after the loop.

````python
count = 1

while count <= 3:
    print("Count:", count)
    count += 1
else:
    print("The loop finished normally.")
````

**Check your understanding**
- Why must `count` change inside the loop?
- What would happen if `count += 1` were removed?
- How would adding `break` affect the `else` block?

````python
# Compare: break prevents the while-else block from running
count = 1

while count <= 3:
    print("Count:", count)
    if count == 2:
        break
    count += 1
else:
    print("This message will not appear.")

print("Program continues after the loop.")
````

## Part 3: `for` Loops and `range()`

A `for` loop is useful when you want to repeat an action for each item in a sequence or a known set of values.

`range()` creates a sequence of integers. Its **stop value is not included**.

| Form | Meaning | Example values |
|---|---|---|
| `range(stop)` | Starts at 0 and stops before `stop` | `range(5)` → 0, 1, 2, 3, 4 |
| `range(start, stop)` | Starts at `start`, stops before `stop` | `range(2, 6)` → 2, 3, 4, 5 |
| `range(start, stop, step)` | Moves by `step` each time | `range(2, 11, 2)` → 2, 4, 6, 8, 10 |

A negative step counts backward. For example, `range(5, 0, -1)` produces 5, 4, 3, 2, 1.

````python
print("1 parameter: range(stop)")
for number in range(5):
    print(number)

print("\n2 parameters: range(start, stop)")
for number in range(2, 6):
    print(number)

print("\n3 parameters: range(start, stop, step)")
for number in range(2, 11, 2):
    print(number)

print("\nCounting backwards")
#make yor code here
````

## Part 4: Loop Flow Control — `break` and `continue`

- `break` ends the **entire loop immediately**.
- `continue` skips the rest of the **current iteration** and moves to the next one.

````python
# break: stop when number reaches 4
for number in range(1, 8):
    if number == 4:
        break
    print(number)

print("The loop stopped.")
````

````python
# continue: skip 4, but keep looping
for number in range(1, 8):
    if number == 4:
        continue
    print(number)

print("The loop completed.")
````

**Predict first:** Which numbers will print in each example? Notice that `break` stops the loop, while `continue` skips only one iteration.

## Part 5: Nested Loops — Rows and Columns

A **nested loop** is a loop inside another loop. The outer loop can represent rows, while the inner loop represents seats in each row.

The example below models a classroom with **3 rows and 4 seats per row**. The inner loop runs completely for each row.

````python
rows = 3
seats_per_row = 4

for row in range(1, rows + 1):
    for seat in range(1, seats_per_row + 1):
        print(f"Row {row}, Seat {seat}")
    print("--- end of row ---")
````

The outer loop runs 3 times. For each row, the inner loop runs 4 times, so there are \(3 \\times 4 = 12\) seat positions.

Try changing `rows` to `4` or `seats_per_row` to `5` and run the cell again.

### Task 4 — `skip_multiples(limit)`

Print numbers from 1 through `limit`, but **skip multiples of 3** by using `continue`. The function should reject a negative or non-integer limit.

````python
def skip_multiples(limit):
  if not isinstance(limit, int) or isinstance(limit, bool) or limit < 1:
        print("Invalid limit: Must be a positive integer.")
        return

  for i in range(1, limit + 1):
        if i % 3 == 0:
            continue
        print(i)

print("Numbers from 1 through 15, skipping multiples of 3:")
print(skip_multiples(15))
````

```python
# Modular Calculator Mini-Program

# add
def add_numbers(num1, num2):
  return num1 + num2

# subtract
def subtract_numbers(num1, num2):
  return num1 - num2

# multiply
def multiply_numbers(num1, num2):
  return num1 * num2

# divide
def divide_numbers(num1, num2):
  if num2 == 0:
    return "Cannot divide by zero."
  else:
    return num1 / num2

first_num = float(input("Enter first number:"))
second_num = float(input("Enter second number:"))

print("-------------")
print("Choose operation:")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")
print("-------------")

choice = input("Enter choice: ")

# check the ser choice and run it
if choice == "1":
  result = add_numbers(first_num, second_num)
  print("\nResult:", result)
elif choice == "2":
    result = subtract_numbers(first_num, second_num)
    print("\nResult:", result)
elif choice == "3":
    result = multiply_numbers(first_num, second_num)
    print("\nResult:", result)
elif choice == "4":
    result = divide_numbers(first_num, second_num)
    print("\nResult:", result)
else:
  print("\nInvalid option. Please choose 1, 2, 3, or 4.")

#Short Reflection

# 1.What functions did you create in your calculator program?
# I created four modular functions: `add_numbers`,
# `subtract_numbers`, `multiply_numbers`, and `divide_numbers`.

# 2.What parameters did your functions use?
#  Each function used two parameters (`num1` and `num2`) to receive
#the numerical values needed for the calculation.

# 3.What arguments were passed when the functions were called?
# The actual numerical values entered by the user and stored in variables
# were passed as arguments into the functions during execution.

# 4.How did your program use the returned value?
# The value returned by each function was captured and stored in a variable
# named `result`, which was then d# isplayed in the console output.

# 5. Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?
# Dividing code into functions makes it organized, readable, and reusable. It also simplifies debugging because individual mathematical operations can
# be tested and fixed independently without disrupting the rest of the application.
```
